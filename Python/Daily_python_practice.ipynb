Character frequency counter
Interview question: Write a Python program to count how many times each character appears in a given string.
Rules
1. Use a for loop.
2. Use a dictionary to store character frequencies.
3. Don't use collections.Counter().
4. Don't manually hardcode the character counts.

```python
company = "mississippi"
cntdic = {}
for c in company:
    if c not in cntdic:
        cntdic[c] = 1
    else:
        cntdic[c] += 1
print(cntdic)
```

Sum of digits of a number
Interview question: Write a Python program to calculate the sum of all digits in a given integer.
Rules
1. Use a while loop.
2. Use the modulus operator % to extract digits.
3. Use integer division // to remove the last digit.
4. Don't convert the number into a string.

```python
number = 4827
total = 0
while number > 0:
    total += number % 10
    number //= 10
print(total)
```

Find the second-largest distinct number
Interview question: Write a Python program to find the second-largest distinct number in a list without sorting it.
Rules
1. Use a for loop.
2. Do not use sort(), sorted(), or max().
3. Handle duplicate numbers correctly.
4. Store the result in a variable and print it.

```python
numbers = [12, 35, 1, 10, 34, 1, 35]
largest = float('-inf')
second_largest = float('-inf')
for num in numbers:
    if num > largest:
        second_largest = largest
        largest = num
    elif largest > num > second_largest:
        second_largest = num
print(second_largest)
```

Move all zeros to the end
Interview question: Given a list of numbers, move all zeros to the end while preserving the original order of the non-zero numbers.
Rules
1. Use a for loop.
2. Do not use sort() or sorted().
3. Preserve the order of non-zero elements.
4. Handle lists containing no zeros or only zeros.
Hint: Think about separating non-zero values from zeros, then combining them.

```python
numbers = [0, 1, 0, 3, 12, 0, 5]
zero = []
for n in numbers:
    if n == 0:
        numbers.remove(n)
        zero.append(n)
numbers.extend(zero)
print(numbers)
```

Aggregate sales by product
Interview question: You receive a list of sales transactions. Write a Python program to calculate the total sales amount for each product using a dictionary.

Rules
1. Use a for loop.
2. Use a dictionary to store the totals.
3. Do not use Counter() or defaultdict().
4. Handle products that appear multiple times.
5. Do not hardcode the product names or totals.

Hint: Use the product name as the dictionary key and its accumulated sales amount as the value.

```python
transactions = [
    ("ITC", 1000),
    ("TCS", 2000),
    ("ITC", 1500),
    ("INFY", 3000),
    ("TCS", 500),
    ("INFY", 1000)
]
total = {}
for stock, amount in transactions:
    if stock not in total:
        total[stock] = 0
    total[stock] += amount
print(total)
    
```

Check whether two strings are anagrams

Interview question: Write a Python program to check whether two strings contain exactly the same characters with the same frequencies, regardless of their order.
Rules
1. Use a for loop.
2. Use dictionaries to compare character frequencies.
3. Do not use sorted() or collections.Counter().
4. Handle strings with different lengths.
5. Treat uppercase and lowercase as equivalent.

```python
str1 = input(str("Enter a string1: ")).lower()
str2 = input(str("Enter a string2: ")).lower()

if len(str1) == len(str2):
    freq1 = {}
    freq2 = {}

    for ch in str1:
        if ch in freq1:
            freq1[ch] += 1
        else:
            freq1[ch] = 1
    for ch in str2:
        if ch in freq2:
            freq2[ch] += 1
        else:
            freq2[ch] = 1
    if freq1 == freq2:
        print("Two Strings are anagrams")
        
    else:
        print("Two Strings are not anagrams")    
else:
    print("Two Strings are not anagrams")
```
