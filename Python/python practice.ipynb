Write Python code to:
1. Find all unique stocks across both sources.
2. Find common stocks.
3. Find stocks available only in source A.
4. Find the total number of unique stocks.
5. Convert the final unique stock collection into a sorted list.

```python
source_a = ["ITC", "TCS", "INFY", "ITC", "HDFC"]
source_b = ["TCS", "WIPRO", "INFY", "SBIN"]
```

```python
#1. Find all unique stocks across both sources.
unique_stk = set(source_a).union(set(source_b))
print(unique_stk)
```

```python
#Find common stocks.
common_stk = set(source_a).intersection(set(source_b))
print(common_stk)
#Find stocks available only in source A.
source_a_only = set(source_a).difference(set(source_b))
print(source_a_only)
#Find the total number of unique stocks.
print(len(unique_stk))
#Convert the final unique stock collection into a sorted list.
print(sorted(unique_stk))
```

Write a for loop to:
1. Print each price.
2. Print only prices greater than 200.
3. Calculate the total of all prices using a loop (don't use sum()).

```python
prices = [100, 200, 150, 300, 250]
```

```python
#Print each price.
for a in prices:
    print(a)
```

```python
#Print only prices greater than 200.
for a in prices:
    if a > 200:
        print(a)
```

```python
#Calculate the total of all prices using a loop (don't use sum()).
total = 0
for a in prices:
    total +=a
print(total)
```

Write a for loop to:
1. Skip invalid transactions (-1) using continue.
2. Stop processing when the transaction amount reaches 400 using break.
3. Print each valid transaction.
4. Calculate the total of the processed transactions.

```python
transactions = [100, 200, -1, 300, 400, -1, 500]
```

```python
total = 0
for a in transactions:
    if a == -1:
        continue
    if a >= 400:
        break
    total+=a
    print("each transaction",a)
print("sum of total processed transaction ",total)
```

This is a useful interview-level exercise.

Write a function named transaction_summary that:
1. Accepts multiple transaction amounts using *args.
2. Accepts employee details using **kwargs.
3. Prints employee details.
4. Calculates and returns the total transaction amount.

```python
def transaction_summary(*args, **kwargs):
    total_amount = sum(args)
    for key,value in kwargs.items():
        print(f"{key}:{value}")
    return f"Total Amount: {total_amount}"
transaction_summary(
    1000, 2000, 3000,
    name="Prabhakaran",
    role="Data Engineer"
)
```

```python
transaction_summary(100, 200, 300, transaction_type="Withdrawal",credit_card="Visa",atm_location="Chennai")
```

```python
result = transaction_summary(1000, 2000, 3000, name="Prabhakaran")
print(result)
```

filter()
* map() transforms every element, whereas filter() selects elements that satisfy a condition.
* Use filter() and lambda to select transactions greater than or equal to 1000.

```python
transactions = [100, 500, 1200, 2500, 800, 4000, 300]
```

```python
fil = list(filter(lambda x:x >= 1000,transactions))
print(fil)
```

Task: Use both filter() and map() to:
1. Select transactions greater than or equal to 1000.
2. Add a 10% processing charge to those selected transactions.
3. Store the final results in a list.

```python
transactions = [100, 500, 1200, 2500, 800, 4000, 300]
```

```python
fil1 = list(filter(lambda x:x >= 1000,transactions))
print(list(map(lambda x: x + x *0.10,fil1)))
```

```python
result = list(map(lambda x: x + x *0.10,filter(lambda x:x >= 1000,transactions)))
print(result)
```
