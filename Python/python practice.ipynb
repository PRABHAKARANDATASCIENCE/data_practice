Exercise 2 – Sets

```python
source_a = [101, 102, 103, 104]
source_b = [103, 104, 105, 106]
```

Write code to find:
1. All unique customer IDs across both sources.
2. Customer IDs present in both sources.
3. Customer IDs present in source A but not source B

```python
set_a = set(source_a)
set_b = set(source_b)
set_c = set_a.union(set_b)
```

```python
print(set_a)
print(set_b)
#All unique customer IDs across both sources.
print(set_c)
```

```python
#Customer IDs present in both sources.
print(set_a.intersection(set_b))
```

```python
#Customer IDs present in source A but not source B
print(set_a.difference(set_b))
```

Exercise 3 – Dictionaries
Write code to:

```python
employee = {
    "id": 101,
    "name": "Arun",
    "department": "IT",
    "salary": 25000
}
```

```python
#Print the employee name.
print(employee["name"])
```
