### Questions (Positional Window Functions & Date Handling)

##### Question 1 (LAG for Price Comparison):

Display each item's order_id, product_id, price, and use the LAG() function to display the price of the previous item (ordered by order_id).

```sql
select order_id,product_id,price,
lag(price) over(order by order_id) as prev_price
from brazilian_e_commerce.sql_practice.olist_order_items_dataset;
```

##### Question 2 (LEAD & Price Difference):

Display each item's order_id, product_id, price, the next item's price using LEAD(), and calculate the difference between the next item price and the current item price.

```sql
select order_id,product_id,price,
lead(price) over (order by order_id) as next_price,
round(next_price-price,2) as price_difference
from brazilian_e_commerce.sql_practice.olist_order_items_dataset;
```

##### Question 3 (Date Truncation & Aggregation):

Using olist_orders_dataset, extract the year and month from order_purchase_timestamp using DATE_TRUNC('month', order_purchase_timestamp) and count the total number of orders placed in each month.

```sql
select date_trunc('month',order_purchase_timestamp) as mon,count(order_id) total_orders
from brazilian_e_commerce.sql_practice.olist_orders_dataset
group by date_trunc('month',order_purchase_timestamp)
order by mon;
```

##### Question 4 (Date Difference / Shipping Lead Time):

Calculate the delivery duration in days for each order in olist_orders_dataset using DATEDIFF(order_delivered_customer_date, order_purchase_timestamp). Display order_id, purchase date, delivery date, and delivery_days. Filter out rows where either date is NULL.

```sql
select order_id,order_purchase_timestamp,order_delivered_customer_date,
date_diff(order_delivered_customer_date,order_purchase_timestamp) as delivery_days
from brazilian_e_commerce.sql_practice.olist_orders_dataset
where order_status='delivered' and order_purchase_timestamp is not null and order_delivered_customer_date is not null;
```
