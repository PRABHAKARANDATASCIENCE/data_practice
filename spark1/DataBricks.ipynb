#### Optimize,z ordering,vaccum

```python
# Create a dummy Delta table to practice on
data = [(i, f"name_{i}", i % 5) for i in range(1000)]
df = spark.createDataFrame(data, ["id", "name", "category"])

# Write as Delta format
df.write.format("delta").mode("overwrite").saveAsTable("practice_delta_table")
```

```python
#using spark sql
spark.sql("OPTIMIZE practice_delta_table")

from delta.tables import DeltaTable
DeltaTable.forName(spark,"practice_delta_table").optimize()
```

```python
#step 3 run optimize with z order
spark.sql("OPTIMIZE practice_delta_table ZORDER BY (category,id)")
```

```python
#step 4 checking table history
spark.sql("DESCRIBE HISTORY practice_delta_table").show(truncate=False)
```

```python
#step 5 runing vaccum (garbage collection)
spark.sql("VACUUM practice_delta_table RETAIN 169 HOURS DRY RUN")
spark.sql("VACUUM practice_delta_table RETAIN 169 HOURS")
```
