```python
df2 = spark.createDataFrame([(1,"Prabhu","Data Engineer"),(2,"Rajesh","Data Analyst"),(3,"Ravi","Data Scientist"), (4,"Revi","Data Analyst"),(5,"ramesh","Data Scientist")], ["id","name","role"])
```

```python
df2.show()
```

```python
df2.withColumn("id", df2.id.cast("string")).show()
df2.printSchema()
```

```python
df2.select("id", "name","role").filter(df2["role"]=="Data Scientist").show()
```

```python
from pyspark.sql.functions import count,when
df2.groupBy("role").agg(count("id")).show()
```

```python
df2=df2.withColumn("salary", when(df2["role"] == "Data Engineer", 90000)
    .when(df2["role"] == "Data Analyst", 70000)
    .when(df2["role"] == "Data Scientist", 120000)
    .otherwise(60000)
)
```

```python
df2.show()
```

```python
from pyspark.sql.window import Window
from pyspark.sql.functions import col, row_number
df2.withColumn("rank",row_number().over(Window.partitionBy("role").orderBy(col("salary").desc()))).show()
```
